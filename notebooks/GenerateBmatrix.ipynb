# Generate a B matrix

Simulated annealing over binary matrices `b` so that the correlations between the
rows of `b` match a target correlation matrix (`?EvoMod.BMatrix.fit_b_matrix`).

```julia
using Pkg
Pkg.activate(@__DIR__)
Pkg.instantiate()

using EvoMod, EvoMod.BMatrix
using LinearAlgebra, Random
Random.seed!(42)
```

```julia
# Parameters
p = 5            # traits
b_2_dims = 100   # columns of b
t_init = 0.001
t_end = 0.0000001
t_steps = 1_000_000
tol = 2

target_matrix = abs.(random_corr(p))
```

```julia
b_matrix, dist = fit_b_matrix(target_matrix; n_cols = b_2_dims, t_init = t_init,
                              t_end = t_end, t_steps = t_steps, tol = tol)
dist
```

```julia
b_projection = ones(p, p)
project!(b_matrix, b_projection)
```

```julia
pcasimilarity(target_matrix, b_projection)
```
